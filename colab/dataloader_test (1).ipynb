```python
def get_word_count(df, df_col_name:str, new_col_name:str):
  df[new_col_name] = df[df_col_name].apply(lambda x:len(str(x).split(" ")))
  return df
```

```python
df = get_word_count(test, "generated-summary", "word-count")
df
```

```python
import spacy
def get_sentence_count(df, df_col_name:str, new_col_name:str):
  nlp = spacy.load('en_core_web_sm')  # بارگذاری مدل زبان انگلیسی
  df[new_col_name] = df[df_col_name].apply(lambda x:len([sent.text for sent in nlp(str(x)).sents]))
  return df
```

```python
t = get_sentence_count(test, "generated-summary", "sentence-count")
t
```
